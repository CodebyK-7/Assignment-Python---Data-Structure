2. Explain the difference between mutable and immutable data types with examples

Ans: Mutable Data Types
Definition: Objects whose values can be changed after they are created.

Examples in Python:

Lists: You can add, remove, or modify elements.

my_list = [1, 2, 3]

my_list[0] = 10   # modifies the first element

print(my_list)    # [10, 2, 3]

Immutable Data Types
Definition: Objects whose values cannot be changed once created. Any modification creates a new object.

Examples in Python:

Strings: Changing a string actually creates a new one.

my_str = "Hello"

my_str = my_str + " World"   # creates a new string

print(my_str)                # "Hello World"



3. What are the main differences between lists and tuples in Python

Ans: Lists vs Tuples:  
Lists are mutable, meaning you can add, remove, or change elements after creation. Tuples are immutable, so once defined, their contents cannot be altered. Lists are better for dynamic collections, while tuples are ideal for fixed data that should remain constant.

4. Describe how dictionaries store data

 Ans:Dictionaries store data as key-value pairs using a hash table. The key is hashed to find its location in memory, making lookups very fast. This structure allows efficient retrieval, insertion, and deletion compared to sequential searches.


5. Why might you use a set instead of a list in Python

Ans: Sets are collections of unique elements, automatically removing duplicates. They are optimized for membership testing, making checks like x in set faster than in lists. Use sets when uniqueness and quick lookups are more important than order.

6. What is a string in Python, and how is it different from a list

Ans: Strings are immutable sequences of characters, meaning they cannot be changed once created. Lists, on the other hand, are mutable and can store elements of any type. Strings are used for text processing, while lists are used for flexible collections.

7. How do tuples ensure data integrity in Python

Ans: Tuples ensure data integrity by being immutable, preventing accidental changes. This makes them useful for storing fixed values like coordinates, dates, or configuration settings. Their immutability guarantees consistency across the program.

8. What is a hash table, and how does it relate to dictionaries in Python ?

Ans: A hash table maps keys to values using a hash function, allowing constant-time access. Python dictionaries are built on hash tables, which is why they are efficient for storing and retrieving data. This makes them powerful for mapping relationships.


9. Can lists contain different data types in Python

ANs: Yes, lists can contain elements of different data types, such as integers, strings, and even other lists. This flexibility makes them versatile for handling mixed data. However, this can sometimes reduce efficiency compared to uniform collections.

10. Explain why strings are immutable in Python

ANs: Strings are immutable because Python stores them in memory in a way that prevents modification. Any change creates a new string object instead of altering the original. This design improves performance and ensures consistency when strings are reused.

11. What advantages do dictionaries offer over lists for certain tasks

ANs: Dictionaries allow fast lookups by key, making them ideal for mapping relationships like names to phone numbers. Lists are better when order matters or when you need to access elements by position. Dictionaries trade order for efficiency in retrieval.

12. Describe a scenario where using a tuple would be preferable over a list

Ans: Tuples are preferable when storing fixed data that should not change, such as geographic coordinates or database keys. Their immutability ensures the values remain consistent. Lists are better when you need flexibility and frequent updates.

13. How do sets handle duplicate values in Python

ANs: Sets automatically remove duplicate values, ensuring all elements are unique. This makes them useful for tasks like filtering repeated entries. Unlike lists, sets do not preserve order, focusing instead on uniqueness and efficiency.

14. How does the “in” keyword work differently for lists and dictionaries

Ans: For lists, the in keyword checks if a value exists among elements. For dictionaries, in checks if a key exists, not the value. This distinction is important when working with mappings versus sequences.


15. Can you modify the elements of a tuple? Explain why or why not

ANs: Tuples cannot be modified because they are immutable. Any attempt to change an element results in an error. This immutability ensures that data stored in tuples remains consistent and secure.

16. What is a nested dictionary, and give an example of its use case

17. Describe the time complexity of accessing elements in a dictionary

ANs: Accessing elements in a dictionary has an average time complexity of O(1). This efficiency comes from the hash table implementation, which allows direct access to values using keys.

18. In what situations are lists preferred over dictionaries

Ans: Lists are preferred when order matters or when you need to store simple sequences. They are also better for tasks involving iteration over elements in a specific order. Dictionaries are more suitable for key-based lookups.

19. Why are dictionaries considered unordered, and how does that affect data retrieval

Ans: Dictionaries are considered unordered because they store data based on hash values rather than sequence. This means retrieval is by key, not position. While modern Python maintains insertion order, the underlying principle is still hash-based.
